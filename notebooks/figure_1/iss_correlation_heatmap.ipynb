# ISS barcode-frequency correlation heatmap

Heatmap of pairwise Pearson correlations between OPS experiments computed on log2 mean-normalised ISS barcode read frequencies — a per-experiment QC measure of barcode-library recovery consistency. Off-diagonal rows/columns that are uniformly low flag experiments with anomalous read distributions.

Input data is the precomputed correlation matrix produced by `ops_monorepo/ops_model/analysis/iss_correlation_heatmap.py`.

The distribution of these pairwise correlations is plotted in `notebooks/SI/figure_S1.ipynb`.

## Imports

```
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
```

## Input path

Precomputed experiment × experiment Pearson correlation matrix from log2 mean-normalised ISS barcode read frequencies, produced by `ops_monorepo/ops_model/analysis/iss_correlation_heatmap.py` and curated into `../../data/figures/figure_1/` (see README).

```
CORR_MATRIX_CSV = Path("../../data/figures/figure_1/iss_barcode_freq_correlation_matrix.csv")
```

## Load correlation matrix

```
corr_df = pd.read_csv(CORR_MATRIX_CSV, index_col=0)
print(f"Loaded {corr_df.shape[0]}x{corr_df.shape[1]} correlation matrix")
```

## Correlation heatmap

Pairwise Pearson correlation between experiments, ordered by OPS experiment number. Colormap floored at 0 (uncorrelated experiments map to white) and capped at 1.

```
corr_values = corr_df.values.astype(float)
n_exp = len(corr_df)

fig, ax = plt.subplots(figsize=(12, 10))
im = ax.imshow(
    corr_values,
    cmap="Reds",
    vmin=0.0,
    vmax=1.0,
    origin="upper",
    aspect="auto",
)

ax.set_xticks(range(n_exp))
ax.set_yticks(range(n_exp))
ax.set_xticklabels(corr_df.columns, rotation=45, ha="right", fontsize=6)
ax.set_yticklabels(corr_df.index, fontsize=6)

plt.colorbar(im, ax=ax, label="Correlation")
ax.set_title(f"ISS barcode-frequency correlation heatmap — {n_exp} experiments")

fig.tight_layout()

output_path = Path("../../output/figure_1/iss_correlation_heatmap.pdf")
output_path.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(output_path, dpi=150, bbox_inches="tight")
plt.show()
```
