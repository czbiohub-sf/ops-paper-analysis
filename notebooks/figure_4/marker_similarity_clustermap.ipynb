# Marker × marker similarity clustermap + per-cluster driver analysis

Two analyses on a shared TF-IDF-normalized marker × geneKO distinctiveness matrix:

1. **Marker × marker Pearson similarity clustermap.** Hierarchically-clustered marker × marker Pearson correlation heatmap. TF-IDF reweighting highlights features hit by few markers, surfacing each marker's unique signal. Rows + columns share the same average-linkage tree on (1 − Pearson r) of similarity rows; a per-marker organelle color strip flanks both axes.
2. **Per-cluster driver analysis (Fig 3D).** Five manually defined marker clusters are scored gene-by-gene with a Cohen-d-like specificity metric. Top drivers are visualized as a vertical overview heatmap, and each cluster's top-25/50/100 drivers are tested for enrichment in GO BP/CC/MF, Reactome, and KEGG via gseapy.enrichr (with the 1,000-gene screen library as background). Enrichment outputs are compiled into per-cluster Excel workbooks (15 sheets each).

All outputs land in `output/figure_4/`. PNG (180 dpi), SVG, and PDF figures preserve text as editable text for downstream figure assembly in Adobe Illustrator.

## Imports

```
import re
import time
from pathlib import Path

import gseapy
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from scipy.cluster.hierarchy import linkage as _linkage
from scipy.spatial.distance import squareform as _squareform

# Illustrator-friendly text exports
plt.rcParams["pdf.fonttype"] = 42      # TrueType in PDF
plt.rcParams["ps.fonttype"] = 42
plt.rcParams["svg.fonttype"] = "none"  # SVG text as <text> elements

FIGURES_DIR = Path("../../output/figure_4")
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
```

## CSV paths

- `DIST_RAW_CSV` — per-gene × per-reporter distinctiveness matrix (genes × reporters), full set including cp + 4i markers. Four 4i markers are dropped at load.
- `METADATA_CSV` — per-marker short name, organelle category, color, and category rank, produced by `data_preprocessing/marker_organelle_metadata.py`.

Both curated into `../../data/figures/figure_4/` (see README).

```
FIGURE_DATA = Path("../../data/figures/figure_4")

DIST_RAW_CSV = FIGURE_DATA / "gene_reporter_distinctiveness_all.csv"
METADATA_CSV = FIGURE_DATA / "marker_organelle_metadata.csv"

EXCLUDED_MARKERS = {"NFkB (4i)", "RSP6 (4i)", "Rb (4i)", "gH2AX (4i)"}
```

## Load distinctiveness matrix and marker metadata

Distinctiveness CSV is genes × reporters; transpose to markers × geneKOs, drop the `all_combined` reference column, drop excluded markers. The metadata CSV is keyed by full marker name and carries `short_name`, `category`, `color`, and `category_rank`.

```
raw = pd.read_csv(DIST_RAW_CSV, index_col=0)
raw = raw.drop(columns=[c for c in raw.columns if c == "all_combined"], errors="ignore")
M = raw.transpose()

excl = {e.lower() for e in EXCLUDED_MARKERS}
M = M.loc[[m for m in M.index if str(m).lower() not in excl]]

meta_df = pd.read_csv(METADATA_CSV)

print(f"markers × geneKOs: {M.shape}")
print(f"metadata rows:     {len(meta_df)}")
```

## TF-IDF normalize, then Pearson similarity

- **TF-IDF**: `tf = value / row_sum`, `idf = log(N / (#markers above column mean))`. Downweights features many markers hit (low IDF) and amplifies rare-hit features.
- **Pearson r**: marker × marker correlation across the normalized feature dimension.

```
# TF-IDF
safe_row = M.sum(axis=1).replace(0, np.nan)
tf = M.div(safe_row, axis=0).fillna(0.0)
df = (M > M.mean(axis=0)).sum(axis=0).replace(0, np.nan)
idf = np.log((M.shape[0] + 1) / (df + 1))
M_norm = tf.mul(idf, axis=1).fillna(0.0)

# Pearson marker × marker similarity
sim = M_norm.transpose().corr(method="pearson")
print(f"similarity shape: {sim.shape}")
```

### Export the TF-IDF matrix (source data for the correlations)

Two shareable tables, for anyone asking for the data behind the marker × marker Pearson r:

- `tfidf_marker_by_gene_matrix.csv` — `M_norm` itself: markers × geneKOs TF-IDF-weighted
  distinctiveness. This is the *only* input to `sim`; each Pearson r is computed between two
  rows of this table.
- `tfidf_per_gene_idf_weights.csv` — the per-gene IDF weight (one scalar per gene) and everything
  that produces it, sorted by IDF descending so rare-hit (up-weighted) genes come first:
  - `n_markers_above_mean` / `markers_above_mean` — the document-frequency count and the actual
    marker names behind it, `;`-joined and ordered most- to least-distinctive for that gene.
    Full marker names, so this column joins against the matrix's row index.
  - `mean_distinctiveness` — the per-gene column mean of `M`, which is also the exact threshold
    that decides membership above, making the count auditable from the table alone.
  - `mean_tfidf` — the gene's mean after reweighting; compare against `mean_distinctiveness` to
    see what the IDF step did to that gene.

Note the gene axis includes `NTC` alongside the library genes, and TF-IDF is a per-(marker, gene)
value — there is no single "per-gene TF-IDF" scalar, which is why the IDF weights are exported
separately.


```
# Source data behind the marker x marker correlations.
# Recomputes the document-frequency / IDF terms from M rather than reusing the
# `df` / `idf` names from the cell above: `df` is reassigned at module scope in
# the enrichment-panel sections below, so this cell stays correct if it is
# re-run out of order.

TFIDF_MATRIX_CSV = FIGURES_DIR / "tfidf_marker_by_gene_matrix.csv"
TFIDF_IDF_CSV = FIGURES_DIR / "tfidf_per_gene_idf_weights.csv"

# 1. The full TF-IDF matrix — the direct input to sim = M_norm.T.corr()
M_norm_out = M_norm.copy()
M_norm_out.index.name = "marker"
M_norm_out.to_csv(TFIDF_MATRIX_CSV)
print(f"wrote {TFIDF_MATRIX_CSV.name}  ({M_norm_out.shape[0]} markers × {M_norm_out.shape[1]} geneKOs)")

# 2. The per-gene IDF weights, with every input that produces them.
#    `mean_distinctiveness` doubles as the audit trail for the IDF count: it is
#    the per-gene column mean of M, i.e. the exact threshold that decides
#    membership in `markers_above_mean`.
gene_mean = M.mean(axis=0)
above_mean = M > gene_mean
markers_above = {
    gene: ";".join(M.loc[above_mean[gene], gene].sort_values(ascending=False).index)
    for gene in M.columns
}
n_above_mean = above_mean.sum(axis=0)

idf_out = pd.DataFrame(
    {
        "n_markers_above_mean": n_above_mean,
        "idf": np.log((M.shape[0] + 1) / (n_above_mean.replace(0, np.nan) + 1)),
        "mean_distinctiveness": gene_mean,
        "mean_tfidf": M_norm.mean(axis=0),
        # ";"-joined full marker names, ordered most- to least-distinctive for
        # this gene. Full names (not the figure's short names) so this column
        # joins directly against the row index of the TF-IDF matrix above.
        "markers_above_mean": pd.Series(markers_above),
    }
)
idf_out.index.name = "gene"
idf_out = idf_out.sort_values("idf", ascending=False)
idf_out.to_csv(TFIDF_IDF_CSV)
print(f"wrote {TFIDF_IDF_CSV.name}  ({len(idf_out)} genes, idf {idf_out['idf'].min():.3f}–{idf_out['idf'].max():.3f})")

# Consistency check: the listed members must match the count they explain.
assert (idf_out["markers_above_mean"].str.count(";") + 1 == idf_out["n_markers_above_mean"]).all(), \
    "markers_above_mean does not match n_markers_above_mean"

idf_out.head()
```

## Clustered heatmap

Average-linkage hierarchical clustering on (1 − Pearson r) of similarity rows; same tree on rows and columns. Color scale floored at 0 so negative correlations clip to white, top anchored at the max off-diagonal value. Organelle color strip on both axes; `*` flags fixed-sample (Cell Painting / 4i) markers.

```
markers = list(sim.index)
marker_meta = meta_df.set_index("marker").loc[markers]
short = marker_meta["short_name"].tolist()

sim_short = sim.copy()
sim_short.index = short
sim_short.columns = short

# Per-marker organelle category color (Series indexed by short name)
cat_short = pd.Series(marker_meta["color"].values, index=short, name="organelle")

# Legend: one entry per category that appears, sorted by display rank
legend_df = (
    marker_meta[["category", "color", "category_rank"]]
    .drop_duplicates(subset="category")
    .sort_values("category_rank")
)
legend_items = list(zip(legend_df["category"], legend_df["color"]))

# Symmetric linkage on (1 - Pearson r) of similarity rows
rows = sim.fillna(0.0).to_numpy(dtype=float)
corr = np.corrcoef(rows)
np.fill_diagonal(corr, 1.0)
dist = np.clip(1.0 - corr, 0.0, None)
dist = (dist + dist.T) / 2.0
np.fill_diagonal(dist, 0.0)
Z = _linkage(_squareform(dist, checks=False), method="average")

# Floor color scale at 0; top = max off-diagonal
sim_arr = sim.to_numpy()
off_diag = sim_arr[~np.eye(sim_arr.shape[0], dtype=bool)]
off_diag = off_diag[np.isfinite(off_diag)]
vmax_use = float(off_diag.max()) if off_diag.size else 1.0
vmin_use = 0.0

side = max(20, len(markers) * 0.35 + 6)
g = sns.clustermap(
    sim_short,
    row_linkage=Z, col_linkage=Z,
    row_cluster=True, col_cluster=True,
    cmap="Reds", vmin=vmin_use, vmax=vmax_use, center=None,
    row_colors=cat_short, col_colors=cat_short,
    figsize=(side, side),
    dendrogram_ratio=(0.10, 0.10),
    colors_ratio=0.02,
    xticklabels=True, yticklabels=True,
    cbar_pos=(0.98, 0.5, 0.018, 0.16),
)
g.ax_heatmap.set_xlabel("")
g.ax_heatmap.set_ylabel("")
plt.setp(g.ax_heatmap.get_xticklabels(), rotation=45, ha="right", va="top", fontsize=17)
plt.setp(g.ax_heatmap.get_yticklabels(), fontsize=17)
if g.cax is not None:
    g.cax.set_ylabel("Pearson r", fontsize=20)
    g.cax.tick_params(labelsize=17)

legend_handles = [
    Patch(facecolor=clr, edgecolor="black", linewidth=0.4, label=cat)
    for cat, clr in legend_items
]
legend = g.fig.legend(
    handles=legend_handles, loc="upper left",
    bbox_to_anchor=(0.98, 0.88), frameon=False, fontsize=20,
    title="Organelle category", title_fontsize=21,
    handlelength=1.6, handleheight=1.6, borderaxespad=0.5,
    labelspacing=0.5,
)
g.fig.canvas.draw()
leg_bbox = legend.get_window_extent().transformed(g.fig.transFigure.inverted())
cbar_h = 0.20
cbar_y = max(leg_bbox.y0 - cbar_h - 0.05, 0.02)
if g.cax is not None:
    g.cax.set_position([0.98, cbar_y, 0.018, cbar_h])

g.fig.suptitle("Marker × marker Pearson r — tfidf (distinctiveness)", fontsize=27, y=1.04)
g.fig.text(
    0.98, 0.005,
    r"$\ast$ indicates a $\mathbf{fixed}$ marker (Cell Painting / 4i)",
    ha="right", va="bottom", fontsize=18,
)

g.fig.savefig(FIGURES_DIR / "clustermap_distinctiveness_pearson_tfidf.svg", bbox_inches="tight")
plt.show()
```

The marker × marker Pearson r matrix exactly as drawn above: short marker names, rows and
columns reordered into the clustermap's dendrogram leaf order (`g.dendrogram_row.reordered_ind`,

```
leaf_order = g.dendrogram_row.reordered_ind          # same tree on rows and columns
sim_clustered = sim_short.iloc[leaf_order, leaf_order]
sim_clustered.index.name = "marker"

CLUSTERMAP_MATRIX_CSV = FIGURES_DIR / "clustermap_distinctiveness_pearson_tfidf_matrix.csv"
sim_clustered.to_csv(CLUSTERMAP_MATRIX_CSV)
print(f"wrote {CLUSTERMAP_MATRIX_CSV}  ({sim_clustered.shape[0]} × {sim_clustered.shape[1]})")

sim_clustered.iloc[:5, :5]
```

## Cluster definitions

Five manually defined marker clusters, guided by hierarchical clustering of the marker × marker Pearson correlation matrix and prior biology. Marker names must match `M_norm.index` exactly. A `sanitize_label` helper collapses spaces, slashes, and other non-alphanumeric characters to `_` so the descriptive dict keys can be reused as filename fragments.

```
# Dict order is load-bearing: it sets the cluster order of the Fig-3D driver
# heatmap (columns and gene row-blocks) and of the enrichment-panel rows.
CLUSTERS = {
    "Mitochondria markers + tubulin": [
        "mitochondria, ChromaLIVE 561 excitation",
        "oxidative stress, CellROX live-cell dye",
        "mitochondria, TOMM70A",
        "microtubules, Tubulin (cp)",
        "mitochondria, TOMM20 (cp)",
    ],
    "Endo-lysosomes": [
        "early endosome, EEA1",
        "endosome, VPS35",
        "lysosome, LAMP1",
        "recycling endosome, TFRC",
        "late endosome, RAB7A",
    ],
    "Cytoskeleton": [
        "actin filament, FastAct_SPY555 Live Cell Dye",
        "f-actin, Phalloidin (cp)",
        "intermediate filaments, VIM",
    ],
    "ER/Golgi markers": [
        "ER/Golgi COP-II, SEC23A",
        "ER/Golgi, COPE",
    ],
    "Nucleus & nucleolus markers": [
        "nucleoli, NPM1 (cp)",
        "nucleolus-DFC, FBL",
        "chromatin, H2BC21",
        "cell proliferation marker, MKI67",
        "nucleolus-GC, NPM3",
        "nuclear speckles, SRRM2",
        "proteasome, PSMB7",
        "laminin, LMNB1",
    ],
}


def sanitize_label(label):
    """Collapse non-alphanumeric runs to '_' so cluster labels are safe in filenames."""
    return re.sub(r"[^A-Za-z0-9]+", "_", label).strip("_")


# Tunable parameters
TOP_N_SIZES = [25, 50, 100]   # driver-list sizes for enrichment
GENE_SETS = [
    "GO_Biological_Process_2023",
    "GO_Cellular_Component_2023",
    "GO_Molecular_Function_2023",
    "Reactome_2022",
    "KEGG_2021_Human",
]
BACKGROUND = M_norm.columns.tolist()  # the screen's 1,000-gene library

# Validate marker names
for label, members in CLUSTERS.items():
    missing = [m for m in members if m not in M_norm.index]
    if missing:
        raise ValueError(f"cluster '{label}' has unknown markers: {missing}")
print(f"All {len(CLUSTERS)} clusters validated. {len(BACKGROUND)}-gene background.")
for label, members in CLUSTERS.items():
    print(f"  {label:34s} n={len(members)}  ->  {sanitize_label(label)}")
```

## Per-cluster Cohen-d-like specificity scoring

For each cluster c and each gene g, compute the cluster-specificity score

$$ d(c, g) = \frac{\bar X_{m\in c, g} - \bar X_{m\notin c, g}}{s_{\text{pooled}}} $$

with pooled SD $s_{\text{pooled}} = \sqrt{((n_{\text{in}}-1)\sigma^2_{\text{in}} + (n_{\text{out}}-1)\sigma^2_{\text{out}})/(n_{\text{in}}+n_{\text{out}}-2)}$ and sample standard deviations (ddof = 1). Scale-free; ranks genes by cluster-vs-background mean shift in pooled-SD units.

```
def cluster_cohen_d(cluster_members):
    """Per-gene Cohen-d-like specificity for one cluster."""
    Xi = M_norm.loc[cluster_members]
    Xo = M_norm.drop(index=cluster_members)
    n_in, n_out = len(Xi), len(Xo)
    std_in = Xi.std(axis=0, ddof=1)
    std_out = Xo.std(axis=0, ddof=1)
    pooled = np.sqrt(
        ((n_in - 1) * std_in ** 2 + (n_out - 1) * std_out ** 2) / (n_in + n_out - 2)
    )
    return (Xi.mean(axis=0) - Xo.mean(axis=0)) / pooled.replace(0, np.nan)


# Compute once; reused by both the heatmap and the enrichment sections below.
cluster_d = {label: cluster_cohen_d(members) for label, members in CLUSTERS.items()}
print("Cohen-d Series computed for all", len(cluster_d), "clusters.")
```

## Top-50 driver gene overview heatmap

Vertical heatmap of the union of the top-50 cluster-specific driver genes per cluster (ranked by the Cohen-d score from section 3), one column per cluster. Genes are grouped into per-cluster blocks in the row order of `CLUSTERS`, with horizontal separators between blocks; the colorbar is capped at d = `VMAX_CAP_LOCAL` (default 5) so small-n and large-n clusters remain visually comparable. Cells with d < 0.05 are masked. Gene labels are omitted because the row count (~200–250) is too dense to read.

```
# Top-50 driver-gene overview heatmap — vertical orientation.
# Union of top-50 per cluster (~200–250 genes total). Gene names are
# omitted since they would be illegible at this row count.

TOP_N_50 = 50
VMAX_CAP_LOCAL = 5.0   # Cohen-d cap for THIS cell only; raise (e.g. 10) to
                       # show more of the tail, lower (e.g. 3) to saturate
                       # earlier and emphasize moderate-effect genes.

# Build deduplicated top-50 union, preserving cluster-block order
gene_order_50 = []
block_boundaries_50 = [0]
for label in CLUSTERS:
    top = (
        cluster_d[label]
        .sort_values(ascending=False)
        .head(TOP_N_50)
        .index.tolist()
    )
    new_genes = [g for g in top if g not in gene_order_50]
    gene_order_50.extend(new_genes)
    block_boundaries_50.append(len(gene_order_50))

# Cluster × gene matrix, then transpose for vertical orientation (genes as rows)
M50 = pd.DataFrame(
    {label: cluster_d[label].loc[gene_order_50] for label in CLUSTERS}
).T
Mv50 = M50.T

print(
    f"Gene-union size: {len(gene_order_50)} "
    f"(no overlaps means {len(CLUSTERS) * TOP_N_50})"
)
print(f"Block boundaries (row indices): {block_boundaries_50}")
print(f"Matrix shape (genes × clusters): {Mv50.shape}")

mask_v50 = Mv50 < 0.05

fig, ax = plt.subplots(
    figsize=(0.2 * Mv50.shape[1] + 3, max(8.0, 0.04 * Mv50.shape[0] + 2))
)
sns.heatmap(
    Mv50,
    mask=mask_v50,
    cmap="Reds",
    vmin=0,
    vmax=VMAX_CAP_LOCAL,
    annot=False,
    linewidths=0,
    cbar_kws={"label": "Cohen-d", "shrink": 0.4},
    ax=ax,
)
cbar = ax.collections[0].colorbar
tick_vals = list(range(0, int(VMAX_CAP_LOCAL) + 1))
cbar.set_ticks(tick_vals)
cbar.set_ticklabels([str(t) for t in tick_vals[:-1]] + [f"≥{int(VMAX_CAP_LOCAL)}"])

# Horizontal separators between the cluster-blocks of genes
for b in block_boundaries_50[1:-1]:
    ax.axhline(b, color="#444", lw=0.8)

ax.set_title(
    f"Top-{TOP_N_50} cluster-specific drivers — Cohen-d across all clusters",
    fontsize=11,
)
ax.set_xlabel("cluster", fontsize=10)
ax.set_ylabel(f"gene KOs (top-{TOP_N_50} union, grouped by source cluster)", fontsize=10)
plt.setp(ax.get_xticklabels(), rotation=45, ha="right", fontsize=9)
ax.set_yticks([])   # gene names omitted — too many rows to read

fig.tight_layout()
base_v50 = FIGURES_DIR / "Fig_3D_top50_drivers_overview_vertical"
fig.savefig(base_v50.with_suffix(".png"), dpi=180, bbox_inches="tight")
fig.savefig(base_v50.with_suffix(".svg"), bbox_inches="tight")
fig.savefig(base_v50.with_suffix(".pdf"), bbox_inches="tight")
plt.show()
print("saved to", base_v50.with_suffix(".png"))
```

## Functional enrichment (Enrichr via gseapy)

For each cluster and each of three driver-list sizes (top-25, top-50, top-100), run gseapy.enrichr against five libraries: GO_BP_2023, GO_CC_2023, GO_MF_2023, Reactome_2022, KEGG_2021_Human. The screen's own 1,000-gene library is used as the background (rather than the human genome) so that the over-representation test is anchored to the actual sampling universe and doesn't over-inflate broadly studied pathways.

Each library is queried individually (gseapy 1.1.x mislabels results when a list of libraries is passed jointly with a custom background) and re-labeled before concatenation. Per-library raw subfolders and concatenated per-(cluster, list-size) CSVs are written, plus a summary table of significant-term counts.

Runtime: 5 clusters × 3 list sizes × 5 libraries = 75 Enrichr API calls; ~2 minutes.

```
def run_enrichr(gene_list, cluster_label, top_n):
    """Enrichr per-library call + concat. Each library queried individually with
    the custom background; the returned Gene_set column is rewritten to the
    correct library name (workaround for gseapy 1.1.x mis-labeling).
    Writes per-library subfolder + concatenated CSV to FIGURES_DIR.
    """
    tag = f"top{top_n}"
    safe_label = sanitize_label(cluster_label)
    frames = []
    for gs in GENE_SETS:
        out_subdir = FIGURES_DIR / f"enrichr_{tag}_{safe_label}" / gs
        res = gseapy.enrichr(
            gene_list=gene_list,
            gene_sets=gs,
            background=BACKGROUND,
            organism="human",
            outdir=str(out_subdir),
            no_plot=True,
        )
        df = res.results.copy()
        df["Gene_set"] = gs  # force correct label
        frames.append(df)
    df = pd.concat(frames, ignore_index=True)
    df.to_csv(FIGURES_DIR / f"enrichment_{tag}_{safe_label}.csv", index=False)
    return df
```

```
summary = []
t0 = time.time()

for label, members in CLUSTERS.items():
    print(f"\n=== {label}  (n={len(members)}) ===")
    d = cluster_d[label].sort_values(ascending=False)
    for n in TOP_N_SIZES:
        gene_list = d.head(n).index.tolist()
        t_start = time.time()
        df = run_enrichr(gene_list, label, n)
        n_sig = int((df["Adjusted P-value"] < 0.05).sum())
        elapsed = time.time() - t_start
        print(f"  top-{n:>3d}  ({len(gene_list)} genes)  →  {n_sig:>3d} sig terms  [{elapsed:.1f}s]")
        summary.append({"cluster": label, "top_n": n, "n_sig_terms": n_sig})

print(f"\nTotal time: {time.time() - t0:.1f}s")

# Pivot to a cluster × list-size table and save.
summary_df = (
    pd.DataFrame(summary)
    .pivot(index="cluster", columns="top_n", values="n_sig_terms")
    .reindex(index=list(CLUSTERS))   # preserve CLUSTERS insertion order
)
summary_df.columns = [f"top-{n} (Adj p<0.05)" for n in summary_df.columns]
summary_csv = FIGURES_DIR / "enrichment_summary.csv"
summary_df.to_csv(summary_csv)
print(f"\nsummary table saved to {summary_csv}")
print("\n=== summary: number of significant enriched terms per (cluster, list size) ===")
print(summary_df.to_string())
```

## Compile enrichment results into per-cluster Excel files

For each marker cluster, build a single `enrichment_<cluster>.xlsx` with **15 sheets** — five pathway/ontology libraries (GO_BP / GO_CC / GO_MF / Reactome / KEGG) × three driver-list sizes (top-25 / top-50 / top-100). Sheets are named `GOBP_top25`, `Reactome_top50`, `KEGG_top100`, etc. Within each sheet, terms are sorted by Adjusted P-value (ascending) and the redundant `Gene_set` column is dropped (the sheet name already encodes it).

Re-uses the per-(cluster, list-size) CSVs written by the enrichment section above, so re-run that section if you change `CLUSTERS` or `GENE_SETS`. Sheets where the corresponding library is missing from the CSV (e.g. you skipped re-running the enrichment section after adding GO_MF) are populated with a placeholder row pointing back here.

## Manuscript enrichment panels

Two publication-ready panels distilled from the per-(cluster, list-size) enrichment CSVs written by the enrichment section above — ported from `data_preprocessing/Fig_3D_cluster_driver_analysis_with_enrichment_panels.ipynb` (sections 9A/9B). Bars show −log10(adj p) of the top-K enriched terms per cluster after Jaccard redundancy pruning.

- **Setup cell** wires up the CSV readers and term-label helpers.
- **All-libraries panel** (`manuscript_enrichment_panel.*`): one row per marker cluster, one column per Enrichr library — all ontologies (GO:BP / GO:CC / GO:MF, Reactome, KEGG).
- **Mix-and-match panel** (`manuscript_enrichment_panel_mix.*`): fixed 2-column layout where each cluster picks its own two libraries. Reuses the selection constants defined by the all-libraries cell.

Each figure writes PNG/SVG/PDF plus a companion `*_terms.csv` to `output/figure_4/`. Run the enrichment section above first so the input CSVs exist.

```
# --- Setup for the manuscript enrichment panel below ---
# Ported from Fig_3D_cluster_driver_analysis_with_enrichment_panels.ipynb
# (section-9 setup + the term-formatting helpers from 9A). Only the readers,
# pruning, and label helpers the mix-and-match panel needs are defined here;
# imports, CLUSTERS, sanitize_label, TOP_N_SIZES, and the enrichment CSVs are
# all produced by the sections above.
import textwrap

# The enrichment loop above wrote the per-(cluster, list-size) CSVs to FIGURES_DIR.
ENRICHMENT_DIR = FIGURES_DIR

# Validate a CSV exists for every (cluster, list_size); otherwise re-run enrichment.
missing = []
for label in CLUSTERS:
    safe = sanitize_label(label)
    for n in TOP_N_SIZES:
        if not (ENRICHMENT_DIR / f"enrichment_top{n}_{safe}.csv").exists():
            missing.append(f"enrichment_top{n}_{safe}.csv")
if missing:
    raise FileNotFoundError(
        f"{len(missing)} expected CSV(s) missing from {ENRICHMENT_DIR}:\n  "
        + "\n  ".join(missing[:6])
        + ("\n  ..." if len(missing) > 6 else "")
        + "\n\nFix: re-run the enrichment section above with the current CLUSTERS."
    )


def load_enrichment(label, top_n):
    """Load one (cluster, list-size) enrichment CSV and parse gene members."""
    safe = sanitize_label(label)
    df = pd.read_csv(ENRICHMENT_DIR / f"enrichment_top{top_n}_{safe}.csv")
    df["gene_members"] = df["Genes"].fillna("").str.split(";").apply(
        lambda gs: frozenset(g for g in gs if g)
    )
    df["n_overlap"] = df["gene_members"].apply(len)
    return df


def jaccard(a, b):
    if not a or not b:
        return 0.0
    return len(a & b) / len(a | b)


def prune_redundant(df_sorted, k, jaccard_thresh=0.5):
    """Greedy: walk down a ranked DataFrame, keep a term unless its gene set
    overlaps a previously kept term by >= jaccard_thresh. Stop at k kept terms."""
    kept_idx, kept_gene_sets = [], []
    for i, row in df_sorted.iterrows():
        if any(jaccard(row["gene_members"], g) >= jaccard_thresh for g in kept_gene_sets):
            continue
        kept_idx.append(i)
        kept_gene_sets.append(row["gene_members"])
        if len(kept_idx) >= k:
            break
    return df_sorted.loc[kept_idx]


# Strip Enrichr-appended GO / Reactome IDs and abbreviate common long phrases
_GO_ID_RE = re.compile(r"\s*\(GO:\d+\)\s*$")
_REACTOME_ID_RE = re.compile(r"\s+R-HSA-\d+\s*$")
_ABBREVIATIONS = [
    (re.compile(r"\bEndoplasmic Reticulum\b", re.IGNORECASE), "ER"),
    (re.compile(r"\bcompartments?\b", re.IGNORECASE),         "compart."),
    (re.compile(r"\bMitochondrial\b"),                        "Mito."),
    (re.compile(r"\bmitochondrial\b"),                        "mito."),
    (re.compile(r"\bMitochondrion\b"),                        "Mito."),
    (re.compile(r"\bmitochondrion\b"),                        "mito."),
    (re.compile(r"\bRibonucleoprotein\b", re.IGNORECASE),     "RNP"),
    (re.compile(r"\bAnterograde\b", re.IGNORECASE),           "antero."),
    (re.compile(r"\bRetrograde\b", re.IGNORECASE),            "retro."),
    (re.compile(r"\bAsparagine\b", re.IGNORECASE),            "Asn"),
    (re.compile(r"\bProtein-Containing\b", re.IGNORECASE),    "Protein"),
]

TERM_WRAP_WIDTH = 22        # chars per line in wrapped term names
TERM_MAX_LENGTH = 60        # cap on total term length before wrapping (… added if exceeded)


def shorten_term(s):
    """Strip trailing GO:nnn / R-HSA-nnn IDs, then apply abbreviations."""
    s = _GO_ID_RE.sub("", s)
    s = _REACTOME_ID_RE.sub("", s)
    for pat, repl in _ABBREVIATIONS:
        s = pat.sub(repl, s)
    return s.strip()


def wrap_term(s):
    """Cap to TERM_MAX_LENGTH, then wrap to TERM_WRAP_WIDTH chars per line.
    `break_on_hyphens=False` keeps compounds like 'ER-Golgi' on one line."""
    if len(s) > TERM_MAX_LENGTH:
        s = s[:TERM_MAX_LENGTH] + "…"
    return textwrap.fill(
        s, width=TERM_WRAP_WIDTH,
        break_long_words=False, break_on_hyphens=False,
    )


PRETTY_LIB = {
    "GO_Biological_Process_2023": "GO:BP",
    "GO_Cellular_Component_2023": "GO:CC",
    "GO_Molecular_Function_2023": "GO:MF",
    "Reactome_2022":              "Reactome",
    "KEGG_2021_Human":            "KEGG",
}

# Discover which libraries are present in the CSVs (handles 4- or 5-library runs).
_probe = load_enrichment(next(iter(CLUSTERS)), TOP_N_SIZES[0])
LIBRARIES = sorted(_probe["Gene_set"].unique().tolist())
print(f"Reading enrichment CSVs from: {ENRICHMENT_DIR}")
print(f"Found {len(LIBRARIES)} libraries: {', '.join(LIBRARIES)}")
```

```
# Manuscript enrichment panel — all libraries (all ontologies).
# Ported from Fig_3D_cluster_driver_analysis_with_enrichment_panels.ipynb (9A).
# Layout: one ROW of subplots per marker cluster, one COLUMN per Enrichr
# library. Bars = -log10(adj p) of the top-K enriched terms after Jaccard
# redundancy pruning. Reads the enrichment CSVs and uses the term-label helpers
# (shorten_term / wrap_term / PRETTY_LIB) from the setup cell above.
#
# The SELECTED_* / JACCARD_THRESHOLD / PADJ_CUTOFF / BAR_COLOR constants set here
# are shared with the mix-and-match panel below, which reuses them.

SELECTED_LIBRARIES = list(LIBRARIES)   # default: every library discovered in setup
SELECTED_LIST_SIZE = 50     # driver-list size to read (top-25 / top-50 / top-100)
SELECTED_K = 2              # bars (terms) shown per subplot
JACCARD_THRESHOLD = 1       # prune a term if it overlaps a kept term by >= this
PADJ_CUTOFF = 0.05
BAR_COLOR = "gray"          # bar fill; any matplotlib color (named or hex)

# Select terms per (library, cluster)
selected = {lib: {} for lib in SELECTED_LIBRARIES}
for lib in SELECTED_LIBRARIES:
    for label in CLUSTERS:
        df = load_enrichment(label, SELECTED_LIST_SIZE)
        sub = df[(df["Gene_set"] == lib) & (df["Adjusted P-value"] < PADJ_CUTOFF)].sort_values("Adjusted P-value")
        selected[lib][label] = prune_redundant(sub, k=SELECTED_K, jaccard_thresh=JACCARD_THRESHOLD)

cluster_list = list(CLUSTERS)
n_rows, n_cols = len(cluster_list), len(SELECTED_LIBRARIES)

# Per-column width: smaller bar area, more visual room for term names
col_width = 2.7 if n_cols <= 3 else 2.4 if n_cols <= 4 else 2.2
# Per-row height: account for multi-line wrapped labels (~1.5 lines avg)
per_row_height = 0.45 * SELECTED_K + 0.55

fig, axes = plt.subplots(
    n_rows, n_cols,
    figsize=(col_width * n_cols + 0.5, per_row_height * n_rows + 0.8),
    squeeze=False,
)

# Per-column x-max (one shared scale per library across all clusters)
col_max = {}
for lib in SELECTED_LIBRARIES:
    vals = [
        -np.log10(selected[lib][label]["Adjusted P-value"]).max()
        for label in cluster_list if not selected[lib][label].empty
    ]
    col_max[lib] = max(vals) if vals else 1.0

for r, label in enumerate(cluster_list):
    for c, lib in enumerate(SELECTED_LIBRARIES):
        ax = axes[r, c]
        kept = selected[lib][label]
        # Keep bar visual height consistent across subplots: always pin K vertical
        # slots, and top-align however many bars survived pruning.
        ax.set_ylim(-0.5, SELECTED_K - 0.5)
        if kept.empty:
            ax.text(0.5, 0.5, "no sig terms", ha="center", va="center",
                    transform=ax.transAxes, color="#888", fontsize=9)
            ax.set_xticks([]); ax.set_yticks([])
            for s in ax.spines.values():
                s.set_visible(False)
        else:
            xvals = -np.log10(kept["Adjusted P-value"].values)
            order = np.argsort(xvals)[::-1]
            terms = kept["Term"].values[order]
            xvals = xvals[order]
            n_overlap = kept["n_overlap"].values[order]
            terms_wrapped = [wrap_term(shorten_term(t)) for t in terms]
            ypos = np.arange(SELECTED_K)[::-1][:len(xvals)]
            ax.barh(ypos, xvals, color=BAR_COLOR, edgecolor="none")
            for y, x, n in zip(ypos, xvals, n_overlap):
                ax.text(x + col_max[lib] * 0.01, y, f" n={n}",
                        va="center", fontsize=7, color="#444")
            ax.set_yticks(ypos)
            ax.set_yticklabels(terms_wrapped, fontsize=7.5)
            ax.set_xlim(0, col_max[lib] * 1.18)
            ax.tick_params(axis="x", labelsize=7)
            ax.spines["top"].set_visible(False)
            ax.spines["right"].set_visible(False)

        # Column titles (library names) on the top row only
        if r == 0:
            ax.set_title(PRETTY_LIB.get(lib, lib), fontsize=11, fontweight="bold", pad=4)

        # x-tick labels + x-label only on bottom row
        if r == n_rows - 1:
            ax.set_xlabel("-log10(adj p)", fontsize=8)
        else:
            ax.tick_params(labelbottom=False)

# Row labels (cluster names) on the left margin, vertical
for r, label in enumerate(cluster_list):
    pos = axes[r, 0].get_position()
    fig.text(0.005, (pos.y0 + pos.y1) / 2, label,
             rotation=90, va="center", ha="left", fontsize=10, fontweight="bold")

fig.suptitle(
    f"Top-{SELECTED_K} enriched terms per marker cluster "
    f"(top-{SELECTED_LIST_SIZE} drivers, Jaccard prune >= {JACCARD_THRESHOLD})",
    fontsize=11, y=0.995,
)
fig.tight_layout(rect=[0.04, 0, 1, 0.99])

base = ENRICHMENT_DIR / "manuscript_enrichment_panel"
fig.savefig(base.with_suffix(".png"), dpi=180, bbox_inches="tight")
fig.savefig(base.with_suffix(".svg"), bbox_inches="tight")
fig.savefig(base.with_suffix(".pdf"), bbox_inches="tight")
plt.show()
print("saved figure to", base.with_suffix(".png"))

# Companion table of selected terms (raw + shortened) for methods / supp
records = []
for lib in SELECTED_LIBRARIES:
    for label in CLUSTERS:
        for rank, (_, row) in enumerate(selected[lib][label].iterrows(), start=1):
            records.append({
                "cluster": label, "library": lib, "rank": rank,
                "Term": row["Term"],
                "Term_short": shorten_term(row["Term"]),
                "Adjusted_P_value": row["Adjusted P-value"],
                "Odds_Ratio": row["Odds Ratio"],
                "n_overlap": row["n_overlap"],
                "Genes": row["Genes"],
            })
selected_csv = ENRICHMENT_DIR / "manuscript_enrichment_panel_terms.csv"
pd.DataFrame(records).to_csv(selected_csv, index=False)
print("term table saved to", selected_csv.name)
```

```
# Manuscript enrichment panel — mix-and-match libraries per cluster.
# Ported from Fig_3D_cluster_driver_analysis_with_enrichment_panels.ipynb (9B).
# Fixed 2-column layout, but each cluster (row) picks its own two libraries.
# Bars = -log10(adj p) of the top-K enriched terms after Jaccard redundancy
# pruning. Reuses SELECTED_LIST_SIZE / SELECTED_K / JACCARD_THRESHOLD /
# PADJ_CUTOFF / BAR_COLOR from the all-libraries panel above and the term-label
# helpers from the setup cell.

COL_WIDTH = 2.5             # per-subplot width in inches (don't drop below ~1.5)

DEFAULT_PAIR = ("GO_Biological_Process_2023", "GO_Cellular_Component_2023")

# Keys must match CLUSTERS exactly — the sanity-check below raises if they drift.
CLUSTER_LIBRARIES = {
    "Mitochondria markers + tubulin":  DEFAULT_PAIR,
    "Endo-lysosomes":                  ("GO_Biological_Process_2023", "KEGG_2021_Human"),
    "Cytoskeleton":                    ("GO_Biological_Process_2023", "Reactome_2022"),
    "ER/Golgi markers":                DEFAULT_PAIR,
    "Nucleus & nucleolus markers":     DEFAULT_PAIR,
}

# Sanity-check the picks against what's in the CSVs
for cluster, libs in CLUSTER_LIBRARIES.items():
    if cluster not in CLUSTERS:
        raise KeyError(f"{cluster!r} is not in CLUSTERS")
    if len(libs) != 2:
        raise ValueError(f"{cluster!r} needs exactly 2 libraries; got {len(libs)}")
    bad = [lib for lib in libs if lib not in LIBRARIES]
    if bad:
        raise ValueError(
            f"{cluster!r}: library/libraries {bad} not in the enrichment CSVs. "
            f"Available: {LIBRARIES}"
        )
# Catch the reverse drift too: a cluster in CLUSTERS with no library pair here.
unassigned = [c for c in CLUSTERS if c not in CLUSTER_LIBRARIES]
if unassigned:
    raise KeyError(f"clusters missing from CLUSTER_LIBRARIES: {unassigned}")

# Select terms for each (cluster, slot)
mix_selected = {}
for cluster, libs in CLUSTER_LIBRARIES.items():
    df = load_enrichment(cluster, SELECTED_LIST_SIZE)
    mix_selected[cluster] = []
    for lib in libs:
        sub = df[(df["Gene_set"] == lib) & (df["Adjusted P-value"] < PADJ_CUTOFF)].sort_values("Adjusted P-value")
        mix_selected[cluster].append(prune_redundant(sub, k=SELECTED_K, jaccard_thresh=JACCARD_THRESHOLD))

cluster_list = list(CLUSTER_LIBRARIES)
n_rows, n_cols = len(cluster_list), 2
per_row_height = 0.45 * SELECTED_K + 0.55

fig, axes = plt.subplots(
    n_rows, n_cols,
    figsize=(COL_WIDTH * n_cols + 0.5, per_row_height * n_rows + 0.8),
    squeeze=False,
)

for r, cluster in enumerate(cluster_list):
    for c in range(n_cols):
        ax = axes[r, c]
        kept = mix_selected[cluster][c]
        lib = CLUSTER_LIBRARIES[cluster][c]
        # Keep bar visual height consistent across subplots: always pin K vertical
        # slots, and top-align however many bars survived pruning.
        ax.set_ylim(-0.5, SELECTED_K - 0.5)

        if kept.empty:
            ax.text(0.5, 0.5, "no sig terms", ha="center", va="center",
                    transform=ax.transAxes, color="#888", fontsize=9)
            ax.set_xticks([]); ax.set_yticks([])
            for s in ax.spines.values():
                s.set_visible(False)
        else:
            xvals = -np.log10(kept["Adjusted P-value"].values)
            order = np.argsort(xvals)[::-1]
            terms = kept["Term"].values[order]
            xvals = xvals[order]
            n_overlap = kept["n_overlap"].values[order]
            terms_wrapped = [wrap_term(shorten_term(t)) for t in terms]
            ypos = np.arange(SELECTED_K)[::-1][:len(xvals)]
            ax.barh(ypos, xvals, color=BAR_COLOR, edgecolor="none")
            x_max = xvals.max()
            for y, x, n in zip(ypos, xvals, n_overlap):
                ax.text(x + x_max * 0.01, y, f" n={n}",
                        va="center", fontsize=7, color="#444")
            ax.set_yticks(ypos)
            ax.set_yticklabels(terms_wrapped, fontsize=7.5)
            ax.set_xlim(0, x_max * 1.18)
            ax.tick_params(axis="x", labelsize=7)
            ax.spines["top"].set_visible(False)
            ax.spines["right"].set_visible(False)
            # Bottom row only gets the x-axis label
            if r == n_rows - 1:
                ax.set_xlabel("-log10(adj p)", fontsize=8)

        # Per-subplot library label (libraries vary per row, so we can't use
        # a single top-row title)
        ax.set_title(PRETTY_LIB.get(lib, lib), fontsize=10, fontweight="bold", pad=4)

# Row labels (cluster names) on the left margin, vertical
for r, label in enumerate(cluster_list):
    pos = axes[r, 0].get_position()
    fig.text(0.005, (pos.y0 + pos.y1) / 2, label,
             rotation=90, va="center", ha="left", fontsize=10, fontweight="bold")

fig.suptitle(
    f"Top-{SELECTED_K} enriched terms - mix-and-match libraries per cluster "
    f"(top-{SELECTED_LIST_SIZE} drivers, Jaccard prune >= {JACCARD_THRESHOLD})",
    fontsize=11, y=0.995,
)
fig.tight_layout(rect=[0.04, 0, 1, 0.99])

base = ENRICHMENT_DIR / "manuscript_enrichment_panel_mix"
fig.savefig(base.with_suffix(".png"), dpi=180, bbox_inches="tight")
fig.savefig(base.with_suffix(".svg"), bbox_inches="tight")
fig.savefig(base.with_suffix(".pdf"), bbox_inches="tight")
plt.show()
print("saved figure to", base.with_suffix(".png"))

# Companion table
records = []
for cluster, libs in CLUSTER_LIBRARIES.items():
    for slot, (lib, df_kept) in enumerate(zip(libs, mix_selected[cluster])):
        for rank, (_, row) in enumerate(df_kept.iterrows(), start=1):
            records.append({
                "cluster": cluster, "slot": slot + 1, "library": lib, "rank": rank,
                "Term": row["Term"],
                "Term_short": shorten_term(row["Term"]),
                "Adjusted_P_value": row["Adjusted P-value"],
                "Odds_Ratio": row["Odds Ratio"],
                "n_overlap": row["n_overlap"],
                "Genes": row["Genes"],
            })
mix_csv = ENRICHMENT_DIR / "manuscript_enrichment_panel_mix_terms.csv"
pd.DataFrame(records).to_csv(mix_csv, index=False)
print("term table saved to", mix_csv.name)
```
